```python
import os
import pandas as pd

# 定义函数计算每种神经元类型的数量及其比例
def calculate_proportions(on_off_df, neuron_types):
    proportions = {}
    for neuron_type in neuron_types:
        neuron_count = len(on_off_df[on_off_df[neuron_type] == 1])
        proportions[neuron_type] = neuron_count
        proportions[f"{neuron_type}_Proportion"] = neuron_count / len(on_off_df.index)
    return proportions

# 定义函数计算神经元对之间的重叠度
def calculate_overlap(on_off_df, neuron_pairs):
    overlap_results = {}
    for pair in neuron_pairs:
        neuron_a, neuron_b = pair
        overlap = len(on_off_df[(on_off_df[neuron_a] == 1) & (on_off_df[neuron_b] == 1)])
        overlap_results[f"{neuron_a}_{neuron_b}_O"] = 2 * overlap / (proportions[neuron_a] + proportions[neuron_b])
        overlap_results[f"{neuron_a}_{neuron_b}_P"] = 2 * proportions[neuron_a] * proportions[neuron_b] / (len(on_off_df.index) * (proportions[neuron_a] + proportions[neuron_b]))
    return overlap_results


# 定义路径和文件名 C:/Users/DELL/Desktop/CLASSIFY-PL2
root_path = 'F:/AAA-ZSY(5)'
IL1_list = ['SYN15-RTIT','SYN17-RTIT','SYN18-ITRT','SYN19-ITRT','SYN20-ITRT','SYN21-ITRT']
session_list = ['session_3_ONOFF.csv','session_6_ONOFF.csv',]  ### Which session do you want to analyze???


# 定义神经元类型和神经元对
neuron_types = ['free_allogroom_ON', 'rear_selfgroom_ON', 'free_allogroom_OFF', 'rear_selfgroom_OFF', 'sniff_ON','sniff_ON']
neuron_pairs = [ ('free_allogroom_ON', 'sniff_ON')]
# neuron_pairs = [ ('free_allogroom_ON', 'sniff_ON'), ('free_allogroom_OFF', 'sniff_OFF')]

results = []

# 遍历每个小鼠和会话，计算比例和重叠度
for mouse_name in IL1_list:
    print('正在处理', mouse_name)
    for session_name in session_list:
        on_off_df = pd.read_csv(os.path.join(root_path, 'Results', mouse_name, mouse_name+'_'+session_name))
        on_off_df.columns = ['Neurons'] + on_off_df.columns[1:].tolist()
        on_off_df = on_off_df.set_index('Neurons')

        proportions = calculate_proportions(on_off_df, neuron_types)
        overlap_results = calculate_overlap(on_off_df, neuron_pairs)

        result = {'mouse_name': mouse_name, 'session': session_name, 'all_neuron': len(on_off_df.index)}
        result.update(proportions)
        result.update(overlap_results)
        results.append(result)

### 将结果保存为Excel文件
save_path = os.path.join(root_path, f"{mouse_name}_ROC_proportion_allsessions.csv")
results_df = pd.DataFrame(results)
results_df.to_csv(save_path)

## results_df.to_excel('Social_Neuron_Degree_Proportion.xlsx')
```
